```python
#Abstraction
from abc import ABC
from abc import abstractmethod
```

```python
class Parent(ABC) :
  @abstractmethod
  def receivedata(self) :
    pass
  @abstractmethod
  def senddata(self) :
    pass
```

```python
class Child(Parent) :
  def receivedata(self) :
    self._data=int(input())
  def senddata(self) :
    return self._data
```

```python
c=Child()
c.receivedata()
print(c.senddata())
```
