```python
CP='C, C++, Python for Competitive Programming\n'
ML='Python, SQL for Machine Learning\n'
WD='HTML, CSS, Bootstrap in Frontend and JavaScript, NodeJS, ReactJS in Backend for Web Development\n'
AD='Kotlin, Java in Native Android and Dart in Flutter and JavaScript in React Native for App Development\n'
Languages=['C\n', 'C++\n', 'Python\n', 'SQL\n', 'Java\n', 'Kotlin\n', 'Dart\n', 'HTML\n', 'CSS\n', 'JavaScript\n']
```

```python
f=open('demo1.txt', 'w')
f.write(CP)
f.write(ML)
f.close()
```

```python
f=open('demo1.txt', 'a')
f.write(WD)
f.write(AD)
f.close()
```

```python
f=open('demo1.txt', 'r')
data=f.read()
print(data)
f.close()
```

```python
f=open('demo1.txt')
while 1 :
  data=f.readline()
  if data=='' :
    break
  print(data, end='')
f.close()
```

```python
try :
  f=open('demo2.txt', 'x')
  f.write('Hello, World!\n')
finally :
  f.close()
```

```python
f=open('demo2.txt', 'r')
print(f.tell())
print(f.read(5))
print(f.tell())
f.seek(7, 0)
print(f.tell())
print(f.read(5))
print(f.tell())
f.seek(0, 2)
print(f.tell())
f.close()
```

```python
with open('demo3.txt', 'w') as f :
  f.writelines(Languages)
with open('demo3.txt', 'r') as f :
  print(f.readlines())
```
