```python
#Functions
def fun(i, f, c, s) :
  print(i+f)
  print(c+s)
fun(33, 0.66, 'S', "udeep")
fun(c='S', i=33, s="udeep", f=0.66)
```

```python
def fun(*T) :
  for i in T :
    print(i, end=' ')
  print()
fun(33)
fun(33, 0.66)
fun(33, 0.66, 'S')
fun(33, 0.66, 'S', "udeep")
```

```python
def fun(**D) :
  for k, v in D.items() :
    print(k, ':', v, end=' ')
  print()
fun(i=33)
fun(i=33, f=0.66)
fun(i=33, f=0.66, c='S')
fun(i=33, f=0.66, c='S', s="udeep")
```

```python
def fun(i, f, c, s) :
  print(i+f)
  print(c+s)
T=(33, 0.66, 'S', "udeep")
fun(*T)
D={'c':'S', 'i':33, 's':"udeep", 'f':0.66}
fun(**D)
```
