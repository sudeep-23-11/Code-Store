```python
import numpy as np
A=np.arange(10, 100, 10)
print(A)
B=np.linspace(10, 100, 10)
print(B)
```

```python
A=np.arange(10, 100, 10)
print(A[1], A[-1])
print(A[:3], A[3:], A[1:4:1])
```

```python
A=np.array([1, 2, 3, 4, 5])
print(np.ndim(A))
print(np.shape(A))
print(A)
B=np.array([[1, 2, 3], [4, 5, 6]])
print(np.ndim(B))
print(np.shape(B))
print(B)
C=np.array([[[1, 2], [3, 4]], [[5, 6], [7, 8]]])
print(np.ndim(C))
print(np.shape(C))
print(C)
```

```python
A=np.zeros((3, 3))
print(A)
B=np.ones((3, 3))
print(B)
C=np.eye(3, 3)
print(C)
D=np.diag([1, 2, 3])
print(D)
E=np.random.rand(3, 3)
print(E)
```

```python
A=np.arange(10, 100, 10)
print(A.dtype)
B=np.arange(10, 100, 10, dtype='float64')
print(B.dtype)
```

```python
A=np.diag([1, 2, 3])
print(A[1, 1])
print(A)
A[1, 1]=5
print(A[1, 1])
print(A)
```

```python
A=np.arange(2, 7, 1)
print(np.sin(A))
print(np.cos(A))
print(np.tan(A))
print(np.log(A))
print(np.log10(A))
print(np.exp(A))
```
