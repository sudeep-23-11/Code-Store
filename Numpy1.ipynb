```python
import numpy as np
```

```python
A=np.arange(10, 100, 10)
B=np.linspace(10, 100, 10)
print(A, B, sep='\n\n')
```

```python
A=np.arange(10, 100, 10)
print(A[1], A[-1])
print(A[:3], A[3:], A[1:4:1])
```

```python
A=np.array([1, 2, 3, 4, 5])
print(np.ndim(A), np.shape(A))
print(A, end='\n\n')
B=np.array([[1, 2, 3], [4, 5, 6]])
print(np.ndim(B), np.shape(B))
print(B, end='\n\n')
C=np.array([[[1, 2], [3, 4]], [[5, 6], [7, 8]]])
print(np.ndim(C), np.shape(C))
print(C, end='\n\n')
```

```python
A=np.zeros((3, 3))
B=np.ones((3, 3))
C=np.eye(3, 3)
D=np.diag([1, 2, 3])
E=np.random.rand(3, 3)
print(A, B, C, D, E, sep='\n\n')
```

```python
A=np.arange(10, 100, 10)
print(A.dtype)
B=np.arange(10, 100, 10, dtype='float64')
print(B.dtype)
```

```python
A=np.diag([1, 2, 3])
A[1, 1]=5
print(A[1, 1])
print(A[1, :])
print(A)
```

```python
A=np.arange(10, 100, 10)
print(A+5, A-5, A*5, A/5, A%5, A//5, A**2, sep='\n\n')
```
