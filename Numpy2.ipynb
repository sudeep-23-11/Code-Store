```python
import numpy as np
A=np.arange(10, 100, 10)
print(A+5)
A=np.diag([1, 2, 3])
print(A*A)
```

```python
A=np.array([1, 2, 3, 4])
B=np.array([1, 3, 2, 4])
print(A==B)
print(np.all(A==B))
print(np.any(A==B))
print(np.array_equal(A, B))
```

```python
A=np.array([0, 0, 1, 1], dtype='bool')
B=np.array([0, 1, 0, 1], dtype='bool')
print(np.logical_not(A))
print(np.logical_and(A, B))
print(np.logical_or(A, B))
print(np.logical_xor(A, B))
```

```python
A=np.array([1, 2, 3, 4, 5])
print(np.sum(A))
print(np.sort(A))
B=np.array([[1, 2, 3], [4, 5, 6]])
print(np.sum(B, axis=0))
print(np.sort(B, axis=0))
print(np.sum(B, axis=1))
print(np.sort(B, axis=1))
```

```python
A=np.array([2, 4, 1, 5, 3, 6])
print(np.argmax(A), np.max(A))
print(np.argmin(A), np.min(A))
```

```python
A=np.array([1, 2, 3, 4, 5])
print(np.mean(A), np.median(A), np.std(A))
B=np.array([[1, 2, 3], [4, 5, 6]])
print(np.mean(B, axis=0), np.median(B, axis=0), np.std(B, axis=0))
print(np.mean(B, axis=1), np.median(B, axis=1), np.std(B, axis=1))
```

```python
B=np.array([[1, 2, 3], [4, 5, 6]])
print(np.ravel(B))
print(np.reshape(B, (6, 1)))
print(B.T)
```
