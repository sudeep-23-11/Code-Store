```python
#Set
S={*range(10, 100, 10)}
for i in S :
  print(i, end=' ')
```

```python
S={*range(10, 100, 10)}
print(10 in S)
print(10 not in S)
```

```python
S1={*range(1, 6, 1)}
S2={6, 7, 8, *S1, 9, 10}
print(S2)
```

```python
S1={*range(1, 6, 1)}
S2=S1
print(S1 is S2)
```

```python
S={1, 2, 0, 4, 5}
print(len(S))
print(max(S), min(S))
print(all(S), any(S))
print(sum(S))
```

```python
S={*range(1, 6, 1)}
S.remove(3)
S.add(0)
print(S)
S.clear()
print(S)
```

```python
S1={*range(1, 11, 1)}
S2={*range(1, 6, 1)}
print(S1.issuperset(S2))
print(S1.issubset(S2))
print(S1.isdisjoint(S2))
```
